### Create a Category that have high and low layoff by using Numpy library
#### Used the filter where for categories the layoff that 
#### If the Total layoff >= 1000 is High and lesser is Low

```python
layoff_data['layoff_category'] = np.where(
    layoff_data['total_laid_off'] >= 1000,
    'High',
    'Low'
)
print(layoff_data['layoff_category'])
```
